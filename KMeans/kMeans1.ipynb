```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data = {
    "Compras": [1, 2, 1.5, 10, 12, 11, 4, 5, 6],
    "Gasto": [2, 3, 2.5, 9, 11, 10, 5, 6, 7]
}

df = pd.DataFrame(data)

print("Dataframe")
print(df)
```

```python
plt.figure(figsize=(6,5))
plt.scatter(df["Compras"], df["Gasto"])
plt.title("Clientes")
plt.xlabel("Compras al mes")
plt.ylabel("Gasto mensual")
plt.grid()
plt.show()
```

```python
inercia = []

for k in range(1, 6):
    modelo = KMeans(n_clusters=k, random_state=42)
    modelo.fit(df)

    inercia.append(modelo.inertia_)

plt.figure(figsize=(6,5))
plt.plot(range(1,6), inercia, marker='o')
plt.title("Metodo del codo")
plt.xlabel("Numero de clusters")
plt.ylabel("Inercia")
plt.grid()
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
df["Cluster"] = kmeans.fit_predict(df)
```

```python
print("\nDataframe con clusters")
print(df)
```

```python
centroides = kmeans.cluster_centers_
print("\nCentroides")
print(centroides)
```

```python
plt.figure(figsize=(7,6))

plt.scatter(
    df["Compras"],
    df["Gasto"],
    c=df["Cluster"],
    s=100
)

plt.scatter(
    centroides[:,0],
    centroides[:,1],
    marker='X',
    s=300
)

plt.title("Clusters de clientes")
plt.xlabel("Compras")
plt.ylabel("Gasto")
plt.grid()
plt.show()
```

Pregunta: ¿Qué tipo de clientes representa cada grupo?

Cluster 1: Clientes con pocas compras y bajo gasto.

Cluster 2: Clientes con compras y gasto medio.

Cluster 3: Clientes frecuentes con alto gasto mensual.
