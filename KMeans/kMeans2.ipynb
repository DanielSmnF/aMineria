```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data = {
    "Ejercicio": [0.5, 1, 1.5, 6, 7, 8, 2, 3, 4],
    "Condicion": [2, 3, 2.5, 8, 9, 10, 4, 5, 6]
}

df = pd.DataFrame(data)

print("Dataframe")
print(df)
```

```python
plt.figure(figsize=(6,5))
plt.scatter(
    df["Ejercicio"],
    df["Condicion"]
)
plt.title("Condicion fisica")
plt.xlabel("Horas de ejercicio")
plt.ylabel("Nivel fisico")
plt.grid()
plt.show()
```

```python
inercia = []
for k in range(1,6):
    modelo = KMeans(n_clusters=k, random_state=42)
    modelo.fit(df)

    inercia.append(modelo.inertia_)

plt.figure(figsize=(6,5))
plt.plot(range(1,6), inercia, marker='o')
plt.title("Metodo del codo")
plt.xlabel("Numero de clusters")
plt.ylabel("Inercia")
plt.grid()
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
df["Cluster"] = kmeans.fit_predict(df)
```

```python
print("\nDataframe con clusters")
print(df)
```

```python
centroides = kmeans.cluster_centers_
print("\nCentroides")
print(centroides)
```

```python
plt.figure(figsize=(7,6))
plt.scatter(
    df["Ejercicio"],
    df["Condicion"],
    c=df["Cluster"],
    s=100
)

plt.scatter(
    centroides[:,0],
    centroides[:,1],
    marker='X',
    s=300
)

plt.title("Clusters de salud fisica")
plt.xlabel("Horas de ejercicio")
plt.ylabel("Condicion fisica")
plt.grid()
plt.show()
```

¿Cómo describirías cada grupo de personas?

Cluster 1: Personas sedentarias con baja condición física.

Cluster 2: Personas con actividad física moderada.

Cluster 3: Personas muy activas con excelente condición física.
