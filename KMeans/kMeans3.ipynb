Ejercicio 3: Agrupamiento de productos según ventas

Objetivo: Clasificar productos según su desempeño en ventas.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
data = {
    "Ventas": [2, 3, 2.5, 15, 18, 20, 5, 6, 7],
    "Ingresos": [3, 4, 3.5, 9, 10, 10, 5, 6, 7]
}

df = pd.DataFrame(data)
print("Dataframe")
print(df)
```

```python
plt.figure(figsize=(6,5))
plt.scatter(
    df["Ventas"],
    df["Ingresos"]
)

plt.title("Productos")
plt.xlabel("Ventas mensuales")
plt.ylabel("Ingresos")
plt.grid()
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)
df["Cluster"] = kmeans.fit_predict(df)
```

```python
print("\nDataframe con clusters")
print(df)
```

```python
centroides = kmeans.cluster_centers_
print("\nCentroides")
print(centroides)
```

```python
plt.figure(figsize=(7,6))
plt.scatter(
    df["Ventas"],
    df["Ingresos"],
    c=df["Cluster"],
    s=100
)

plt.scatter(
    centroides[:,0],
    centroides[:,1],
    marker='X',
    s=300
)

plt.title("Clusters de productos")
plt.xlabel("Ventas")
plt.ylabel("Ingresos")
plt.grid()
plt.show()
```

Pregunta: ¿Qué tipo de productos representa cada cluster?

Cluster 1: Productos con bajas ventas y bajos ingresos.

Cluster 2: Productos con desempeño medio.

Cluster 3: Productos estrella con altas ventas y altos ingresos.
