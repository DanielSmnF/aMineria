***Ejercicio 3 — Rendimiento académico***

Contexto: Una escuela quiere identificar tipos de estudiantes según desempeño. (Promedio , asistencia)

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
```

```python
X = np.array([
    [95, 98],  # Alumno A
    [92, 94],  # Alumno B
    [88, 90],  # Alumno C
    [65, 70],  # Alumno D
    [60, 68],  # Alumno E
    [58, 62],  # Alumno F
    [30, 40],  # Alumno G
    [35, 38]   # Alumno H
])

alumnos = [
    'A', 'B', 'C',
    'D', 'E', 'F',
    'G', 'H'
]
```

```python
plt.figure(figsize=(7,6))
plt.scatter(X[:,0], X[:,1], s=100)
plt.title('Datos de rendimiento academico')
plt.xlabel('Promedio')
plt.ylabel('Asistencia')
plt.grid(True)
plt.show()
```

```python
Z = linkage(X, method='ward')
```

```python
plt.figure(figsize=(10,5))
dendrogram(
    Z,
    labels=alumnos
)

plt.title('Dendrograma de estudiantes')
plt.xlabel('Alumnos')
plt.ylabel('Distancia')
plt.show()
```

```python
modelo_2 = AgglomerativeClustering(
    n_clusters=2,
    linkage='ward'
)

clusters_2 = modelo_2.fit_predict(X)
plt.figure(figsize=(7,6))
plt.scatter(
    X[:,0],
    X[:,1],
    c=clusters_2,
    s=100
)

for i, txt in enumerate(alumnos):
    plt.annotate(txt, (X[i,0], X[i,1]))

plt.title('Agrupamiento con 2 clusters')
plt.xlabel('Promedio')
plt.ylabel('Asistencia')
plt.grid(True)
plt.show()
```

```python
print("clusters con 2 grupos:")
for alumno, cluster in zip(alumnos, clusters_2):
    print(alumno, "-> cluster", cluster)
```

```python
modelo_3 = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

clusters_3 = modelo_3.fit_predict(X)
plt.figure(figsize=(7,6))
plt.scatter(
    X[:,0],
    X[:,1],
    c=clusters_3,
    s=100
)

for i, txt in enumerate(alumnos):
    plt.annotate(txt, (X[i,0], X[i,1]))

plt.title('Agrupamiento con 3 clusters')
plt.xlabel('Promedio')
plt.ylabel('Asistencia')
plt.grid(True)
plt.show()
```

```python
print("Clusters con 3 grupos:")
for alumno, cluster in zip(alumnos, clusters_3):
    print(alumno, "-> cluster", cluster)
```

**¿Qué alumnos se parecen más?**

A y B, D y E, y G y H porque tienen valores muy parecidos.

**¿Qué perfiles de estudiantes aparecen?**

Alto rendimiento,
Rendimiento medio,
Bajo rendimiento

**¿Qué decisión institucional podría tomarse?**

Dar apoyo a alumnos con bajo rendimiento y reconocer a los destacados.

**¿Qué ocurre cuando el número de clusters cambia?**

Con mas clusters hay grupos más específicos y detallados.

**¿Qué grupo requeriría más intervención académica?**

El grupo de G y H por sus bajas calificaciones y asistencia.
